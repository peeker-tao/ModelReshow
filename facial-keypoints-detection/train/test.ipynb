```python
image_str = df.Image[0]
points = df.iloc[0].drop("Image").values.astype(np.float32)
```

```python
def parse_image(image_str,points = None, size=(96, 96)):
    arr = np.array(image_str.split(" "), dtype=np.float32).reshape(size)/255.0
    plt.imshow(arr, cmap='gray')
    if points is not None:
        for i in range(0, len(points), 2):
            x, y = points[i], points[i+1]
            plt.scatter(x, y, c='red', s=20)
    plt.show()
    return arr
```

```python
parse_image(image_str, points)
```

```python
def parse_image_id(image_id,df1,df2,size=(96, 96)):
    image_str = df1[df1.ImageId == image_id].Image.values[0]
    points = df2[df2.ImageId == image_id].drop("ImageId", axis=1).values.flatten().astype(np.float32)
    arr = np.array(image_str.split(" "), dtype=np.float32).reshape(size)/255.0
    plt.imshow(arr, cmap='gray')
    if points is not None:
        for i in range(0, len(points), 2):
            x, y = points[i], points[i+1]
            if x > 0 and y > 0:  # 只绘制有效点
              plt.scatter(x, y, c='red', s=20)
    plt.show()
    
```

```python
def parse_image_(image_id,df,size=(96, 96)):
    image_str = df.Image[image_id]
    points = df.iloc[image_id].drop("Image").values.astype(np.float32)
    parse_image(image_str, points)
```
